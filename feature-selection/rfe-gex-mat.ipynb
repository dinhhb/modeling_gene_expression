```python
import polars as pl

def is_many_to_one(df, col_a, col_b):
    """Checks if each value in A maps to EXACTLY ONE value in B."""
    return (
        df.group_by(col_a)
        .agg(pl.col(col_b).n_unique().alias("unique_count"))
        .select(pl.col("unique_count") <= 1)
        .to_series()
        .all()
    )

def is_one_to_many(df, col_a, col_b):
    """Checks if each value in B maps to EXACTLY ONE value in A."""
    return is_many_to_one(df, col_b, col_a)

def check_relationship(df, col_a, col_b):
    # Using a slightly cleaner syntax for Python 3.10+
    m_to_1 = is_many_to_one(df, col_a, col_b)
    o_to_m = is_one_to_many(df, col_a, col_b)
    
    if m_to_1 and o_to_m:
        return "1-to-1"
    elif m_to_1:
        return "Many-to-One"
    elif o_to_m:
        return "One-to-Many"
    else:
        return "Many-to-Many"

# Test it now
print(check_relationship(gex, "patientID", "sample_id"))
```
