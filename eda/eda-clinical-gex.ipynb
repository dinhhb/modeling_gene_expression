```python
import polars as pl

BINARY_PFS = True
RNA_SEQ_ONLY = True

pfs_dir = "binary" if BINARY_PFS else "three-classes"
rna_suffix = "_rna" if RNA_SEQ_ONLY else ""

clinical_gex = pl.read_csv(f'../dataset/created/{pfs_dir}/clinical_gex{rna_suffix}.csv')
clinical_gex
```

```python
CLINICAL_COLS = ["patientID", "sex", "age", "AJCC_stage", "M_stage", "LDH", "pfs_label",
                "immunotherapy_treatment", "pre_MAPKi_treatment",
                "source", "cobimetinib", "dabrafenib", "trametinib","vemurafenib", 
                "C195W", "K601I", "MND", "R558Q", "V600E", "V600K","V600R"]
GEX_COLS = [c for c in clinical_gex.columns if c not in CLINICAL_COLS]
```

```python
gex_long = (
    clinical_gex.select(['source'] + GEX_COLS)
    .unpivot(on=GEX_COLS, index='source', variable_name='gene', value_name='value')
    .with_columns(pl.col('value').cast(pl.Float64, strict=False))
    .drop_nulls('value')
)

source_stats = gex_long.group_by("source").agg([
    pl.col("value").mean().alias("mean"),
    pl.col("value").median().alias("median"),
    pl.col("value").std().alias("std"),
    pl.col("value").min().alias("min"),
    pl.col("value").max().alias("max")
])
print(source_stats)
```

```python
import matplotlib.pyplot as plt
import numpy as np

sources_sorted = sorted(gex_long['source'].unique().to_list())
n = len(sources_sorted)

data_by_source = [
    np.clip(gex_long.filter(pl.col('source') == s)['value'].to_numpy(), 1e-5, None)
    for s in sources_sorted
]

fig, ax = plt.subplots(figsize=(8, 6))

bp = ax.boxplot(data_by_source, patch_artist=True,
                medianprops=dict(color='red', linewidth=2),
                flierprops=dict(marker='.', markerfacecolor='steelblue', markersize=1, alpha=0.3))

for patch in bp['boxes']:
    patch.set_facecolor('steelblue')
    patch.set_alpha(0.6)

ax.set_yscale('log')
ax.set_xticks(range(1, n + 1))
ax.set_xticklabels(sources_sorted, rotation=15, ha='right')
ax.set_ylabel('Expression value (log scale, clipped at 1e-5)')
ax.set_title('Expression value distribution by source')
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

sources_sorted = sorted(gex_long['source'].unique().to_list())
n = len(sources_sorted)

data_by_source = [
    np.log1p(np.clip(gex_long.filter(pl.col('source') == s)['value'].to_numpy(), 0, None))
    for s in sources_sorted
]

method_colors = {
    'Hugo et al.':  '#4C72B0',
    'Kwong et al.': '#4C72B0',
    'Yan et al.':   '#4C72B0',
    'Rizos et al.': '#DD8452',
    'Long et al.':  '#DD8452',
}
colors = [method_colors[s] for s in sources_sorted]

fig, ax = plt.subplots(figsize=(8, 6))

parts = ax.violinplot(data_by_source, positions=range(1, n + 1),
                      showmedians=False, showextrema=False, bw_method=0.15)

for pc, color in zip(parts['bodies'], colors):
    pc.set_facecolor(color)
    pc.set_alpha(0.6)

for i, (pos, color) in enumerate(zip(range(1, n + 1), colors)):
    ax.boxplot([data_by_source[i]], positions=[pos],
               widths=0.12, patch_artist=True, showfliers=False,
               medianprops=dict(color='white', linewidth=2),
               boxprops=dict(facecolor=color, alpha=0.5),
               whiskerprops=dict(color=color, linewidth=1.2),
               capprops=dict(color=color, linewidth=1.2))

legend_handles = [
    plt.matplotlib.patches.Patch(color='#4C72B0', label='RNA seq'),
    plt.matplotlib.patches.Patch(color='#DD8452', label='Microarray'),
]
ax.legend(handles=legend_handles, title='Method', fontsize=9)

ax.set_ylim(top=12)
ax.set_xticks(range(1, n + 1))
ax.set_xticklabels(sources_sorted)
ax.set_ylabel('log1p(Expression value)')
# ax.set_title('Distribution of log1p expression values by source')
plt.tight_layout()
plt.show()
```

```python
pfs_counts = (
    clinical_gex
    .select(['source', 'pfs_label'])
    .group_by(['source', 'pfs_label'])
    .agg(pl.len().alias('count'))
    .sort(['source', 'pfs_label'])
)

pfs_totals = pfs_counts.group_by('source').agg(pl.col('count').sum().alias('total'))
pfs_counts = pfs_counts.join(pfs_totals, on='source').with_columns(
    (pl.col('count') / pl.col('total') * 100).alias('pct')
)

sources = sorted(pfs_counts['source'].unique().to_list())
labels = sorted(pfs_counts['pfs_label'].unique().to_list())
colors = ['red', 'orange', 'green']
label_names = {0: 'Short', 1: 'Intermediate', 2: 'Long'}

x = np.arange(len(sources))
width = 0.8 / len(labels)

fig, ax = plt.subplots(figsize=(7, 5))

for i, label in enumerate(labels):
    subset = pfs_counts.filter(pl.col('pfs_label') == label)
    count_map = dict(zip(subset['source'].to_list(), subset['count'].to_list()))
    pct_map = dict(zip(subset['source'].to_list(), subset['pct'].to_list()))
    offset = (i - len(labels) / 2 + 0.5) * width

    counts = [count_map.get(s, 0) for s in sources]
    pcts = [pct_map.get(s, 0.0) for s in sources]
    bars = ax.bar(x + offset, counts, width, label=label_names[label], color=colors[i], alpha=0.8)

    for bar, pct in zip(bars, pcts):
        ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 0.3,
                f'{pct:.0f}%', ha='center', va='bottom', fontsize=8)

ax.set_xticks(x)
ax.set_xticklabels(sources, rotation=15, ha='right')
ax.set_ylabel('Count')
# ax.set_title('pfs_label distribution by source')
ax.legend(title='pfs_label')
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

if BINARY_PFS:
    label_names = {0: "Short", 1: "Long"}
    colors = {"Short": "#e74c3c", "Long": "#27ae60"}
    response_order = ["Short", "Long"]
else:
    label_names = {0: "Short", 1: "Intermediate", 2: "Long"}
    colors = {"Short": "#e74c3c", "Intermediate": "#f39c12", "Long": "#27ae60"}
    response_order = ["Short", "Intermediate", "Long"]

plot_df = (
    clinical_gex
    .select(["age", "pfs_label"])
    .drop_nulls()
    .with_columns(pl.col("pfs_label").cast(pl.Int32))
    .to_pandas()
)
plot_df["Response"] = plot_df["pfs_label"].map(label_names)

bins = [0, 40, 55, 70, 120]
labels_bins = ["<40", "40–55", "55–70", "70+"]
plot_df["age_group"] = pd.cut(plot_df["age"], bins=bins, labels=labels_bins, right=False)

responses_present = [v for v in response_order if v in plot_df["Response"].unique()]

grouped = (
    plot_df.groupby(["age_group", "Response"], observed=True)
    .size()
    .unstack(fill_value=0)
    .reindex(columns=responses_present)
)
grouped_pct = grouped.div(grouped.sum(axis=1), axis=0) * 100
totals = grouped.sum(axis=1)

fig, ax = plt.subplots(figsize=(7, 5))

bottom = np.zeros(len(grouped_pct))
for resp in responses_present:
    ax.bar(grouped_pct.index.astype(str), grouped_pct[resp], bottom=bottom,
           label=resp, color=colors[resp], alpha=0.85)
    for i, (pct, b, cnt) in enumerate(zip(grouped_pct[resp], bottom, grouped[resp])):
        if pct > 6:
            ax.text(i, b + pct / 2, f"{cnt}\n({pct:.0f}%)", ha="center", va="center",
                    fontsize=8, color="white", fontweight="bold")
    bottom += grouped_pct[resp].values

for i, total in enumerate(totals):
    ax.text(i, 102, f"n={total}", ha="center", va="bottom", fontsize=8)

ax.set_ylim(0, 112)
ax.set_xlabel("Age group")
ax.set_ylabel("Proportion (%)")
ax.legend(title="PFS Label", fontsize=8)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

rna_seq_sources = {'Hugo et al.', 'Kwong et al.', 'Yan et al.'}
microarray_sources = {'Rizos et al.', 'Long et al.'}

label_names = {0: 'Short', 1: 'Inter', 2: 'Long'}
label_order = [0, 1, 2]

counts = {}
for lbl in label_order:
    subset = clinical_gex.filter(pl.col('pfs_label') == lbl)
    counts[lbl] = {
        'RNA-seq':    subset.filter(pl.col('source').is_in(list(rna_seq_sources))).height,
        'Microarray': subset.filter(pl.col('source').is_in(list(microarray_sources))).height,
    }

x = np.arange(len(label_order))
width = 0.35
colors = {'RNA-seq': '#4C72B0', 'Microarray': '#DD8452'}

fig, ax = plt.subplots(figsize=(6, 5))

for i, method in enumerate(['RNA-seq', 'Microarray']):
    vals = [counts[lbl][method] for lbl in label_order]
    bars = ax.bar(x + (i - 0.5) * width, vals, width, label=method,
                  color=colors[method], alpha=0.8)
    for bar, v in zip(bars, vals):
        ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 0.3,
                str(v), ha='center', va='bottom', fontsize=9)

ax.set_xticks(x)
ax.set_xticklabels([label_names[l] for l in label_order])
ax.set_ylabel('Count')
ax.set_xlabel('PFS Label')
ax.legend(title='Method')
plt.tight_layout()
plt.show()
```
