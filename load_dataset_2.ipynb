## Helper functions

```python
import pandas as pd
import numpy as np
```

```python
def count_unique_value_per_class(df):
    for feat in df.columns.tolist():
        print(f'{df[feat].value_counts()}\n')
        
def count_null_per_class(df):
    print(f'{df.isnull().sum()}')
```

## SNV dataset

```python
snvs_df = pd.read_csv("dataset/snvs.csv")
snvs_df
```

```python
count_unique_value_per_class(snvs_df)
```

```python
snvs_df.drop(columns=['id', 'creation_datetime', 'HGVSp', 'other_prelevements', 'uniprot_id', 'source'], inplace=True)
snvs_df.isnull().sum()
```

```python
snvs_df = snvs_df[snvs_df.temporality == 'pre treatment'].reset_index(drop=True)
snvs_df
```

```python
gene_list = [
    "BRAF", "RAC1", "STK11", "PIK3CA", "PDGFRA", "ERBB4", "FGFR3", "NRAS", 
    "ALK", "DDR2", "AKT1", "KIT", "JAK2", "MET", "FGFR1", "EGFR", 
    "CDKN2A", "GNA11", "ERBB2", "CTNNB1", "TERT", "FGFR2", "MAP2K1"
]

snvs_df = snvs_df[snvs_df['HGNC'].isin(gene_list)].reset_index(drop=True)
snvs_df
```

```python
from sklearn.preprocessing import OneHotEncoder, MultiLabelBinarizer

categorical_cols = ['HGNC', 'HGVSp_short', 'variant_type']

for feat in categorical_cols:
    snvs_df[feat] = snvs_df[feat].fillna('Unknown')

enc = OneHotEncoder(sparse_output=False)

one_hot_encoded = enc.fit_transform(snvs_df[categorical_cols])
one_hot_snvs_df = pd.DataFrame(one_hot_encoded, columns=enc.get_feature_names_out(categorical_cols))

snvs_df_encoded = pd.concat([snvs_df['patientID'], one_hot_snvs_df], axis=1)
# display(snvs_df_encoded)


snvs_df['consequence_list'] = snvs_df['consequence'].fillna('Unknown').str.split(',')

mlb = MultiLabelBinarizer()
enc_arr = mlb.fit_transform(snvs_df['consequence_list']) 

snvs_df_mlb = pd.DataFrame(enc_arr, columns=['cons_' + name for name in mlb.classes_])

snvs_df_encoded = pd.concat([snvs_df_encoded, snvs_df_mlb], axis=1)
snvs_df_encoded
```

```python
snvs_df_encoded = snvs_df_encoded.groupby(['patientID']).max().reset_index()
snvs_df_encoded
```

## Merge clinical+snvs

```python
clinical_snvs_df = clinical_df.merge(snvs_df_encoded, how='left', on='patientID')
clinical_snvs_df
```

```python
count_unique_value_per_class(clinical_snvs_df)
```
