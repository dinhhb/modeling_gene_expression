This code reads the preprocessed Gene expression data and merge it with the cleaned Clinical dataset

Set up some configurations

```python
BINARY_PFS = False       # True: binary classification, False: 3-class classification
RNA_SEQ_ONLY = False     # True: RNA-seq samples only, False: RNA-seq + microarray samples

pfs_dir = "binary" if BINARY_PFS else "three-classes"
rna_suffix = "_rna" if RNA_SEQ_ONLY else ""
```

Read the Gene expression matrix (gex) from MelanoDB

```python
import pandas as pd
import numpy as np

gex = pd.read_csv('../dataset/created/gex.csv')

print(gex.shape)
gex.head()
```

Read the cleaned Clinical dataset

```python
clinical = pd.read_csv(f"../dataset/created/{pfs_dir}/clinical{rna_suffix}.csv")
clinical = clinical.drop(['PFS_month', 'brain_metastasis'], axis=1)

print(clinical.shape)
clinical.head()
```

Merge the two datasets

```python
# There are DUPLICATED patientID (126 patients, 143 patients-samples)
gex_wide = gex.pivot_table(index=['patientID', 'sample_id'], columns='HGNC', values='value', aggfunc='mean')

print(gex_wide.shape)
gex_wide.head()
```

```python
clinical_gex = clinical.merge(gex_wide.reset_index(), on='patientID', how='inner')

print(clinical_gex.shape)
clinical_gex.head()
```
