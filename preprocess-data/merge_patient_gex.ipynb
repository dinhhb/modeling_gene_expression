```python
import polars as pl
```

# Patient (Clinical)

```python
patient_df = pl.read_csv("dataset/clinical.csv")
patient_df
```

```python
def count_unique_value_per_class(df):
    for feat in df.columns:
        print(f'{df[feat].value_counts()}\n')
        
count_unique_value_per_class(patient_df)
```

```python
patient_df = patient_df.drop(pl.col(['id', 'creation_datetime', 'original_patientID', 'OS_status', 'OS_month', 'source', 'brain_metastasis', 'immunotherapy_treatment', 'M_stage', 'CNA_data', 'SNV_data', 'GEX_data']))
patient_df
```

```python
gex_df = pl.read_csv("gex_data/gex.csv")
merge_df = gex_df.join(patient_df, on='patientID', how='inner')
merge_df
```

```python
merge_df.select(pl.all().null_count())
```

```python
threshold = merge_df.height * 0.5
cols_to_keep = [col.name for col in merge_df.iter_columns() if col.null_count() <= threshold]
merge_df = merge_df.select(cols_to_keep)
```

```python
merge_df
```

```python
pl.Config.set_tbl_rows(-1)   # show all rows

merge_df.select([
    pl.col("patientID"),
    pl.sum_horizontal(pl.all().is_null()).alias("null_count")
])
```

```python
merge_df = merge_df.filter(
    pl.sum_horizontal(pl.all().is_null()) <= merge_df.width * 0.5
)
merge_df
```

```python
merge_df.write_csv('gex_data/gex_patient.csv')
```
