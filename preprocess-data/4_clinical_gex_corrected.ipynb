# Merge Clinical + corrected GEX data

```python
import pandas as pd
```

```python
gex_mat = pd.read_csv(f"../dataset/created/gex_mat_corrected_sig.csv").set_index('sample_id').T
gex_mat['sample_id'] = gex_mat.index
gex_mat
```

```python
clinical_gex = pd.read_csv(f"../dataset/created/clinical_gex.csv", index_col='Unnamed: 0')
clinical_gex
```

```python
CLINICAL_COLS = ["patientID", "sex", "age", "AJCC_stage", "M_stage", "LDH", 'sample_id', 'pfs_label',
                "immunotherapy_treatment", "pre_MAPKi_treatment",
                "source", "cobimetinib", "dabrafenib", "trametinib","vemurafenib", 
                "C195W", "K601I", "MND", "V600E", "V600K","V600R"]
GEX_COLS = [c for c in clinical_gex.columns if c not in CLINICAL_COLS]

clinical_gex = clinical_gex[CLINICAL_COLS]
```

```python
gex_mat_reset = gex_mat.reset_index()
clinical_gex_1 = clinical_gex.merge(gex_mat_reset, on='sample_id', how='inner')

clinical_gex_1
```

```python
clinical_gex_1.to_csv(f"../dataset/created/clinical_gex_corrected.csv", index=False)
```
