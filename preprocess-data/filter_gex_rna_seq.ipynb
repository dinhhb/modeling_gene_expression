```python
gex_df = gex_df.filter(pl.col('temporality') == 'pre treatment')
gex_df
```

```python
# Hugo et al., Kwong et al. and Yan et al. studies because they used a similar method 
# (mRNA expression profiling performed with Illumina HiSeq 2000)
source_rna_seq = ['doi:10.1016/j.cell.2015.07.061', 'doi:10.1158/1078-0432.CCR-18-0720', 'doi:10.1172/JCI78954DS1']

gex_df = gex_df.filter(pl.col('source').is_in(source_rna_seq))
gex_df
```

```python
gex_df = gex_df.drop(['id', 'creation_datetime', 'GeneID', 'description', 'temporality'])
gex_df.select(pl.all().null_count())
```

```python
gex_df = gex_df.drop_nulls(subset=['patientID', 'HGNC'])
gex_df.select(pl.all().null_count())
```

```python
gex_df
```

```python
source_map = {
    'doi:10.1016/j.cell.2015.07.061': 'Hugo et al.',
    'doi:10.1172/JCI78954DS1': 'Kwong et al.',
    'doi:10.1158/1078-0432.CCR-18-0720': 'Yan et al.'
}

gex_df = gex_df.with_columns(pl.col('source').replace(source_map))
gex_df
```
