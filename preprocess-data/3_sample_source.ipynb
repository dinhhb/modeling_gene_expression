# 3. Create sample-source metadata

## Read GEX_MAT

```python
import polars as pl
gex_mat = pl.read_csv(f'../dataset/created/gex_mat.csv')
gex_mat
```

# Read GEX

```python
gex = pl.read_csv(f"../dataset/created/gex.csv")
gex
```

## Fitler GEX col -> sample_source

```python
sample_source = (
    gex
    .select(["sample_id", "source", 'patientID'])
    .unique()
)
sample_source
```

```python
sample_source['patientID'].value_counts(sort=True)  # There are duplicated patientIDs
```

## Read Clinical for getting Target class

```python
clinical = pl.read_csv(f'../dataset/created/clinical.csv')
clinical
```

## Reorder sample_source sampleIDs according to gex_mat sampleIDs

```python
# Get column order from gex_mat (excluding HGNC)
col_order = gex_mat.columns[1:]  # skip HGNC

# Reorder sample_source rows to match gex_mat column order
sample_source_reordered = (
    pl.DataFrame({"sample_id": col_order})
    .join(sample_source, on="sample_id", how="left")
)


sample_source_reordered
```

## Add Target class

```python
sample_source_reordered = (
    sample_source_reordered
    .join(
        clinical.select(["patientID", "pfs_label"]),
        on="patientID",
        how="left"
    )
).drop_nulls(pl.col('pfs_label'))
sample_source_reordered
```

## Filter GEX_MAT to have same samples as sample_source

```python
valid_samples = sample_source_reordered["sample_id"].to_list()
cols_to_keep = ["HGNC"] + [c for c in gex_mat.columns if c in valid_samples]
gex_mat_filtered = gex_mat.select(cols_to_keep)

gex_mat_filtered
```

```python
sample_source_reordered.write_csv(f"../dataset/created/sample_source.csv")
gex_mat_filtered.write_csv(f"../dataset/created/gex_mat.csv")
```
