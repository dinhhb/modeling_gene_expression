```python
import polars as pl
```

```python
# gex_mat = pl.read_csv("dataset/gene_expressions_matrix.csv")
gex_mat = pl.read_csv('../dataset/created/gex_mat_rna_seq.csv')
gex_mat
```

```python
sample_cols = [c for c in gex_mat.columns if c != "HGNC"]

# Cast only the String-typed sample columns to Float64
string_sample_cols = [c for c in sample_cols if gex_mat[c].dtype == pl.String]
print(f"String cols to fix: {string_sample_cols}")

gex_mat = gex_mat.with_columns(
    pl.col(string_sample_cols).cast(pl.Float64, strict=False)
)

# Verify
print(gex_mat.select(string_sample_cols).dtypes)
```

```python
gex_mat_sort_by_var = (
    gex_mat.with_columns(
        pl.concat_list(sample_cols)
        .list.drop_nulls()
        .map_elements(lambda x: x.var(), return_dtype=pl.Float64)
        .alias("_variance")
    )
    .sort("_variance", descending=True)
    # .head(500)
    # .drop("_variance")
)
gex_mat_sort_by_var
```

```python
gex = pl.read_csv("dataset/gene_expressions.csv")
gex
```

```python
sample_source = (
    gex.select(["sample_id", "source"])
    .unique()
    # .filter(pl.col('source').is_in(['doi:10.1016/j.cell.2015.07.061', 'doi:10.1172/JCI78954DS1', 'doi:10.1158/1078-0432.CCR-18-0720']))
)
sample_source
```

```python
# Reorder columns to match sample_source row order
sample_order = sample_source["sample_id"].to_list()
gex_mat_reordered = gex_mat_sort_by_var.select(["HGNC"] + sample_order)

gex_mat_reordered
```

```python
gex_mat_reordered = gex_mat_reordered.head(500)
gex_mat_reordered
```

```python
gex_mat_reordered.write_csv("dataset/gene_expressions_matrix_clean.csv")
sample_source.write_csv("dataset/sample_source.csv")
```
