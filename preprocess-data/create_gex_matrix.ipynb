```python
import polars as pl
```

```python
DATASET = 'gex_pre'
```

```python
gex_df = pl.read_csv(f"../dataset/created/gex/{DATASET}.csv")
gex_df
```

```python
dupes = gex_df.to_pandas().duplicated(subset=['HGNC', 'sample_id'], keep=False)
print(gex_df.to_pandas()[dupes])
```

```python
print(f"Shape before deduplication: {gex_df.shape}")
```

```python
# For duplicate (HGNC, sample_id) pairs, keep the one with the max value
gex_df = gex_df.sort('value', descending=True).unique(subset=['HGNC', 'sample_id'], keep='first')
print(f"Shape after deduplication: {gex_df.shape}")
```

```python
dupes = gex_df.to_pandas().duplicated(subset=['HGNC', 'sample_id'], keep=False)
print(gex_df.to_pandas()[dupes])
```

```python
gex_df
```

```python
gex_matrix = gex_df.pivot(on='sample_id', index='HGNC', values='value')
gex_matrix
```

```python
gex_matrix.write_csv(f'../dataset/created/gex_mat/{DATASET}.csv')
```
