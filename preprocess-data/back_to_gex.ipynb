```python
import polars as pl

DATASET = 'gex_pre'

gex_mat = pl.read_csv(f"../dataset/created/gex_mat_corrected/{DATASET}.csv")
gex_mat
```

```python
sample_source = pl.read_csv(f"../dataset/created/sample_source/{DATASET}.csv")
sample_source
```

```python
gex_info = gex_mat.transpose(column_names='HGNC').insert_column(
    0, pl.Series('sample_id', gex_mat.columns[1:])
)
gex_info
```

```python
gex_complete = sample_source.join(other=gex_info, on='sample_id')
gex_complete
```

```python
gex_complete = gex_complete.remove(pl.col('source') == 'Louveau et al.')
gex_complete
```

```python
gex_complete.write_csv(f"../dataset/created/gex_corrected/{DATASET}.csv")
```
