```python
import pandas as pd
```

```python
DATASET = 'gex_pre'
```

```python
sample_source = pd.read_csv(f"../dataset/created/sample_source/{DATASET}.csv")
sample_source
```

```python
clinical_gex = pd.read_csv(f"../dataset/created/clinical_gex/clinical_gex_w_target.csv")
clinical_gex = clinical_gex[['patientID', 'pfs_label']]
clinical_gex
```

```python
clinical_gex[clinical_gex.duplicated('patientID', keep=False)].sort_values('patientID')
```

```python
ids_source = set(sample_source['patientID'])
ids_clinical = set(clinical_gex['patientID'])

overlap = ids_source & ids_clinical
only_source = ids_source - ids_clinical
only_clinical = ids_clinical - ids_source

print(f"sample_source patients:  {len(ids_source)}")
print(f"clinical_gex patients:   {len(ids_clinical)}")
print(f"Overlap:                 {len(overlap)}")
print(f"Only in sample_source:   {len(only_source)}")
print(f"Only in clinical_gex:    {len(only_clinical)}")
```

```python
merged = sample_source.merge(clinical_gex, how='outer', on='patientID')
merged
```
